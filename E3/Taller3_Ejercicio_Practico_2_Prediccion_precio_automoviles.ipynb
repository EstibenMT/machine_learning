## Contexto de negocio

A partir de especificaciones técnicas, dimensiones y atributos de distintos vehículos, se busca **predecir el precio de venta del automóvil (`price`)**. Como la variable objetivo es un valor numérico continuo (dinero) y no una categoría, este es un problema de **regresión**, a diferencia del Ejercicio 1 (Heart Disease) que era de clasificación.

## Preparación de los datos separados

Aquí **no podemos usar `y` directamente**, porque `y = automobile.data.targets` trae `symboling` (el target oficial de UCI), y no `price`. Por eso trabajaremos únicamente sobre `X` (que sí contiene las 25 columnas, incluyendo `price`), y más adelante extraeremos `price` manualmente para construir nuestra propia variable objetivo.

Trabajamos sobre una copia (`features_df`) para no modificar los datos originales devueltos por `fetch_ucirepo`.

```python
import pandas as pd

features_df = X.copy()

features_df.head()
```

### Revisión de las variables de entrada

Revisamos la estructura de `features_df` para conocer cuántos registros tiene, qué tipo de datos maneja cada columna y si existen valores faltantes.

```python
features_df.info()
```

```python
features_df.isna().sum()
```

### Revisión de la variable objetivo (`price`)

Como `price` viene dentro de `features_df` (no en un `target_df` separado), la revisamos directamente ahí. Es importante confirmar si tiene valores faltantes, porque una variable objetivo debe estar completa antes de entrenar el modelo.

```python
features_df['price'].describe()
```

```python
features_df['price'].isna().sum()
```

### Eliminación de registros sin variable objetivo

A diferencia de las demás variables de entrada (que se imputan dentro del pipeline para evitar fuga de datos), los registros que **no tienen valor de `price`** no sirven para entrenar ni para evaluar el modelo, porque no hay nada contra qué comparar la predicción. Por eso, estos registros se eliminan antes de la partición, no se imputan.

```python
registros_validos = features_df['price'].notna()

features_df_completo = features_df[registros_validos].reset_index(drop=True)

features_df_completo.shape
```

## Construcción de la matriz de características (X_model) y el vector objetivo (y_model)

Ahora sí construimos nuestra propia variable objetivo: `y_model` es la columna `price`, y `X_model` son **todas las demás columnas**, eliminando `price` de las features.

```python
y_model = features_df_completo['price'].copy()
X_model = features_df_completo.drop(columns=['price']).copy()

X_model
```

```python
y_model
```

## Partición en conjuntos de entrenamiento y prueba

Dividimos los datos en un conjunto de entrenamiento (80%) y uno de prueba (20%). A diferencia del Ejercicio 1 (clasificación), aquí **no se usa `stratify`**, porque ese parámetro solo aplica quando la variable objetivo es categórica (para mantener la proporción de clases); como `price` es una variable numérica continua, no tiene sentido estratificar por ella.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_model, y_model, test_size=0.2, random_state=917)
```

## Verificación del tipo de dato de cada columna

Antes de construir el pipeline de preprocesamiento, verificamos con `dtypes` qué tipo de dato tiene cada columna. **las columnas de tipo `object` (texto) van al pipeline categórico**; **las columnas numéricas (`int64`/`float64`) van al pipeline numérico**.

```python
X_model.dtypes
```

### Identificación automática de columnas numéricas y categóricas

Como este dataset tiene 24 columnas de entrada en `X_model` (25 originales menos `price`, que ahora es nuestro target), aquí conviene usar la técnica automática con `select_dtypes`.

```python
columnas_numericas = X_model.select_dtypes(include=['int64', 'float64']).columns
columnas_numericas
```

```python
columnas_categoricas = X_model.select_dtypes(include=['object']).columns
columnas_categoricas
```

## Definición de transformaciones para variables numéricas y categóricas

Para las variables numéricas se aplica imputación por mediana (varias columnas como `normalized-losses`, `bore`, `stroke`, `horsepower` o `peak-rpm` tienen valores faltantes, y la mediana es robusta a valores atípicos) seguida de escalado estándar, necesario para que el modelo trate todas las variables en una escala comparable.

Para las variables categóricas se aplica imputación por el valor más frecuente (por ejemplo `num-of-doors` tiene un par de valores faltantes) seguida de codificación One-Hot, eliminando la primera categoría para evitar colinealidad.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

pipeline_numerico = Pipeline(steps=[('imputacion_numerica', SimpleImputer(strategy='median')), ('escalado_numerico', StandardScaler())])
```

```python
from sklearn.preprocessing import OneHotEncoder

pipeline_categorico = Pipeline(steps=[('imputacion_categorica', SimpleImputer(strategy='most_frequent')), ('vdummies', OneHotEncoder(drop='first', handle_unknown='ignore'))])
```

## Integración de los transformadores en un ColumnTransformer

Despues de la separación es necesario combinar ambos pipelines, porque el dataset tiene tanto columnas `object` (categóricas) como numéricas.

```python
from sklearn.compose import ColumnTransformer

preprocesador = ColumnTransformer(transformers=[('num', pipeline_numerico, list(columnas_numericas)), ('cat', pipeline_categorico, list(columnas_categoricas))])
```

## Construcción y entrenamiento del pipeline final con Linear Regression

El modelo seleccionado es **Linear Regression**, ya que el objetivo es predecir un valor numérico continuo (el precio del vehículo).

```python
from sklearn.linear_model import LinearRegression

automobile_pipeline = Pipeline(steps=[('preprocesamiento', preprocesador), ('regresor', LinearRegression())])

automobile_pipeline.fit(X_train, y_train)
```

## Predicción y evaluación del modelo

Generamos las predicciones sobre el conjunto de prueba y calculamos el **RMSE** (Root Mean Squared Error), la métrica estándar para evaluar modelos de regresión.

```python
y_pred = automobile_pipeline.predict(X_test)

y_pred
```

```python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(y_test, y_pred)
rmse
```

### Interpretación del resultado

El modelo de regresión lineal obtuvo un **RMSE de 2631,02 dólares**.

Esto significa que las predicciones del precio se alejan del valor real aproximadamente en **2631 dólares**, aunque el error puede ser mayor o menor dependiendo del vehículo.

Para interpretar mejor este resultado, lo comparamos con los precios del conjunto de datos:

- Precio promedio: **13 207,13 dólares**.
- Precio mínimo: **5118 dólares**.
- Precio máximo: **45 400 dólares**.
- Desviación estándar: **7947,07 dólares**.

El RMSE representa aproximadamente el **19,9 % del precio promedio**. Esto indica que el modelo logra realizar predicciones razonables, pero todavía presenta diferencias importantes en algunos vehículos, especialmente cuando sus precios son muy altos o tienen características poco comunes.

En conclusión, el modelo aprendió una relación útil entre las características técnicas del automóvil y su precio. Sin embargo, el error de 2631 dólares muestra que la regresión lineal no logra explicar completamente todas las variaciones del precio.
